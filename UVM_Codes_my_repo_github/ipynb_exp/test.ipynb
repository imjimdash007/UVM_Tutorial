```python
%%writefile hello_world.v
module hello_world;
    initial begin
        $display("Hello, World! Welcome to vsim inside Jupyter.");
        $finish;
    end
endmodule
```

```python
%%bash
# 1. Create a work library (ModelSim/QuestaSim requirement)
vlib work

# 2. Compile the Verilog file using vlog
vlog hello_world.v

# 3. Run the simulation using vsim in command-line/batch mode (-c)
vsim -c -do "run -all; quit" hello_world
```
